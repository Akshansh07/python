### Wordnet Lemmatizer 

Lemmatization Technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word rather than root stem, the output of stemming. after lemmatization, we will be getting a valid word that means the same thing.

NLTK provide WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the 
WordNet CorpusReader classto find a lemma. Let us understand it with an example - 

## basic eaxampe Q&A, Chatbots, text summaritazation.

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatizer.lemmatize('going')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()


''' pos - Noun - n
    verb - v
    adjective - a
    adverb - r 
'''
lemmatizer.lemmatize('going', pos='v')
```

```python
words = ["eating", "eat", "eaten", "going", "go", "gone", "Writing", "write", "written", "Programming", "program", "programmed", "History", "historical", "Historically", "History"]
```

```python
for word in words:
    print(word+"--->"+ lemmatizer.lemmatize(word, pos='v'))
```
