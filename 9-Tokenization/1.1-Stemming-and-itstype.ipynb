### Stemming

Stemming is the process of reducing word to its word stem that affixes to suffixes or the roots of words known as a lemma. Stemming is imporatant in natural language understanding (NLU) and language processing (NLP).

```python
## Classification problem
## Comment of product is a positive review or negative review
## Review ------> eating, eat, eaten [going, go, gone] --> go   {allare comment menning so steam word is eat this stem find the root 
## word of all the words}

words = ["eating", "eat", "eaten", "going", "go", "gone", "Writing", "write", "written", "Programming", "program", "programmed", "History", "historical", "Historically", "History"]
```

```python
## PorterStemmer

from nltk.stem import PorterStemmer

stemming = PorterStemmer()

for word in words:
    print(word+"--->"+ stemming.stem(word))
```

```python
stemming.stem('congratulation')
```

```python
stemming.stem("sitting")
```

### RegexStemmer class

NLTK has RegexStemmer class with the help of which we can easily implement Regular Expression Stemming algorithms. It basically takes a single regular expression and remove any prefix or suffics that matches the expression. Let us see an example

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4) 

## if we place ing$ doller in last then it remove ing from the word from the last. if we remove doller it remove ing from last or first from both side.

reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

snowballsstemmer = SnowballStemmer("english")

for word in words:
    print(word+"---->"+ snowballsstemmer.stem(word))
```
