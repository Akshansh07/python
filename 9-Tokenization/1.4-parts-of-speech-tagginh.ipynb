### Part of speech tags


```python
## Speech of Dr apj abdul kalam

paragraph = """Good morning to the respected teachers and my dear friends.
 I am very happy to be here to deliver a speech on the topic “My Vision for India”. 
 I am very thankful to the organizers for giving me this opportunity. 
 I am also thankful to all of you for sparing your valuable time to listen to me. 
 I am very happy to see all of you here. I am sure that you are all very eager to listen to my speech. 
So, without wasting any time, let me start my speech.
"""
```

```python
# from nltk.stem import PorterStemmer
import nltk
from nltk.corpus import stopwords

# stemmer = PorterStemmer() # Creating an object of the PorterStemmer class.

sentences = nltk.sent_tokenize(paragraph) # Tokenizing the paragraph into sentences.
```

```python
sentences
```

```python
import nltk

nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')  # For newer NLTK versions
```

```python
## We will find the pos tag

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word not in set(stopwords.words('english'))]
    # sentences[i] = ' '.join(words) # Converting all the list of words into sentences.
    pos_tags=nltk.pos_tag(words) # Finding the pos tag of each word in the sentence.
    print(pos_tags)
```
